```python
# Import the specific Tukey HSD function
import pandas as pd
from statsmodels.stats.multicomp import pairwise_tukeyhsd
df = pd.read_csv('Neg-Geotaxis-2.csv')

# Run the test
# 'Ratio' is the data points, 'Group' is the names of the groups
tukey_results = pairwise_tukeyhsd(endog=df['Ratio'], groups=df['Group'], alpha=0.05)

# Prints the pre-formatted results table

print(tukey_results)
```
